```python
# Which shipping mode has the biggest problem?

Shipping_analysis= (
    df.groupby('Shipping Mode')
        .agg(
            orders = ('Order Id', 'nunique'),
            late_rate=("Late_delivery_risk", "mean"),
            avg_actual_days=("Days for shipping (real)", "mean"),
            avg_scheduled_days=("Days for shipment (scheduled)", "mean")

        ) 
)

Shipping_analysis['late_rate'] *= 100

Shipping_analysis.sort_values(
    'late_rate', ascending= False
)
```

```python
monthly_delivery = (
    df.groupby(["year", "month"])
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_shipping_days=("Days for shipping (real)", "mean")
      )
      .reset_index()
)

monthly_delivery["late_rate"] *= 100

monthly_delivery.head(20)
```

```python
monthly_delivery.sort_values("late_rate", ascending=False).head(10)
```

```python
yearly_delivery = (
    df.groupby("year")
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_shipping_days=("Days for shipping (real)", "mean")
      )
      .reset_index()
)

yearly_delivery["late_rate"] *= 100

yearly_delivery
```

```python
region_analysis = (
    df.groupby("Order Region")
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_shipping_days=("Days for shipping (real)", "mean")
      )
      .reset_index()
)

region_analysis["late_rate"] *= 100

region_analysis.sort_values(
    "late_rate",
    ascending=False
)
```

```python
market_analysis = (
    df.groupby("Market")
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_shipping_days=("Days for shipping (real)", "mean")
      )
      .reset_index()
)

market_analysis["late_rate"] *= 100

market_analysis.sort_values(
    "late_rate",
    ascending=False
)
```

```python
mode_region = (
    df.groupby(["Shipping Mode", "Order Region"])
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_shipping_days=("Days for shipping (real)", "mean"),
          avg_scheduled_days=("Days for shipment (scheduled)", "mean")
      )
      .reset_index()
)

mode_region["late_rate"] *= 100

mode_region.sort_values(
    "late_rate",
    ascending=False
).head(20)
```

```python
gap_by_mode = (
    df.groupby("Shipping Mode")["shipping_gap"]
      .agg(["mean", "median", "min", "max"])
      .reset_index()
)

gap_by_mode
```

```python
delivery_status = (
    df.groupby("Delivery Status")
      .agg(
          records=("Order Id", "count"),
          orders=("Order Id", "nunique"),
          sales=("Sales", "sum"),
          profit=("Order Profit Per Order", "sum")
      )
      .reset_index()
)

delivery_status
```

```python
delivery_profit = (
    df.groupby("Delivery Status")
      .agg(
          orders=("Order Id", "nunique"),
          sales=("Sales", "sum"),
          avg_profit=("Order Profit Per Order", "mean")
      )
      .reset_index()
)

delivery_profit
```

```python
category_analysis = (
    df.groupby("Category Name")
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          sales=("Sales", "sum"),
          avg_profit=("Order Profit Per Order", "mean")
      )
      .reset_index()
)

category_analysis["late_rate"] *= 100

category_analysis.sort_values(
    "late_rate",
    ascending=False
).head(15)
```

```python
mode_category = (
    df.groupby(["Shipping Mode", "Category Name"])
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_gap=("shipping_gap", "mean")
      )
      .reset_index()
)

mode_category["late_rate"] *= 100

mode_category.sort_values(
    "late_rate",
    ascending=False
).head(20)
```

```python
customer_segment= (
    df.groupby('Customer Segment')
    .agg(
        order=('Order Id', 'nunique'),
        late_rate=('Late_delivery_risk', 'mean'),
        sales=('Sales', 'sum'),
        profit=('Order Profit Per Order', 'sum')
    )
    .reset_index()
)

customer_segment['late_rate'] *= 100

customer_segment.sort_values('late_rate', ascending=False)
```

```python
financial_impact = (
    df.groupby("Delivery Status")
      .agg(
          orders=("Order Id", "nunique"),
          sales=("Sales", "sum"),
          profit=("Order Profit Per Order", "sum"),
          avg_profit=("Order Profit Per Order", "mean")
      )
      .reset_index()
)

financial_impact
```

```python
late_orders = df[df["Late_delivery_risk"] == 1]

late_sales = late_orders["Sales"].sum()
total_sales = df["Sales"].sum()

late_sales_share = late_sales / total_sales * 100

print("Late-order sales:", late_sales)
print("Total sales:", total_sales)
print("Late-order sales share:", late_sales_share)
```

```python
country_analysis = (
    df.groupby("Order Country")
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_gap=("shipping_gap", "mean")
      )
      .reset_index()
)

country_analysis["late_rate"] *= 100

country_analysis.sort_values(
    "late_rate",
    ascending=False
).head(20)
```

```python
root_cause_summary = (
    df.groupby(["Shipping Mode", "Order Region"])
      .agg(
          orders=("Order Id", "nunique"),
          late_rate=("Late_delivery_risk", "mean"),
          avg_gap=("shipping_gap", "mean"),
          sales=("Sales", "sum"),
          profit=("Order Profit Per Order", "sum")
      )
      .reset_index()
)

root_cause_summary["late_rate"] *= 100

root_cause_summary = root_cause_summary.sort_values(
    ["late_rate", "orders"],
    ascending=[False, False]
)

root_cause_summary.head(20)
```
